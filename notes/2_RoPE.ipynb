```python
import torch

dim = 6

# 1. 生成偶数序列：[0, 2, 4]
# 这里的 0, 2, 4 其实对应公式里的 2i (i是组的索引)
idx = torch.arange(0, dim, 2).float() 
print("idx:", idx) # 输出: tensor([0., 2., 4.])

# 2. 计算频率 (对应的公式是 10000 ^ (-2i/dim))
# 算出来分别是: 10000^0, 10000^(-2/6), 10000^(-4/6)
inv_freq = 1.0 / (10000 ** (idx / dim))
print("inv_freq:", inv_freq) 
# 输出大概是: tensor([1.0000, 0.0464, 0.0022]) 
# 这就是 3 个小组各自的“旋转速度”。
```

```python
max_seq_len = 5

# 3. 生成所有可能的位置：[0, 1, 2, 3, 4]
t = torch.arange(max_seq_len).float()
print("t (位置):", t) # 输出: tensor([0., 1., 2., 3., 4.])

# 4. 用“位置”乘以“频率”，算出每个 token 每一组的绝对角度
# 注意：t 形状是 [5]，inv_freq 形状是 [3]
# torch.outer 会把它们两两相乘，变成一个 [5, 3] 的矩阵
freqs = torch.outer(t, inv_freq) 
print("freqs (角度表):\n", freqs)
```

```python
# 5. 算 cos 和 sin
cos_table = freqs.cos()
sin_table = freqs.sin()

# 6. 魔法操作 repeat_interleave
# 作用：把 [A, B, C] 变成 [A, A, B, B, C, C]
cos_table = cos_table.repeat_interleave(2, dim=-1)
sin_table = sin_table.repeat_interleave(2, dim=-1)

print("cos_table 形状:", cos_table.shape) # 输出: [5, 6]
print("cos_table :\n", cos_table)
print("cos_table 形状:", sin_table.shape) # 输出: [5, 6]
print("cos_table :\n", sin_table)
```

```python
# 前面就是说先进行对位置进行编码，和你的输入时没有关系的，先根据你的max_len(这个是你输入的)和dim(token的维度是6)生成一个RoPE的旋转位置编码表
# 假设输入的形状是 [1, 5, 6]：表示bacth_size是1，token是5，token的维度是6
x = torch.randn(1, 5, 6)
print('x：',x)
# 7. 奇偶拆分
x_even = x[..., 0::2] # 取第 0, 2, 4 位。形状变成 [1, 5, 3]
print("x_even形状：",x_even.shape)
print("x_even：\n",x_even)
x_odd  = x[..., 1::2] # 取第 1, 3, 5 位。形状变成 [1, 5, 3]
print("x_odd形状：",x_odd.shape)
print("x_odd：\n",x_odd)
```

```python
# 8. 取出对应的 cos 和 sin（我们也只要偶数位，刚好 3 个一循环）
# 为了能跟 x 匹配，在最前面加个维度，变成 [1, 5, 3]
cos_half = cos_table[..., 0::2].unsqueeze(0)
print("cos_half形状：",cos_half.shape)
print("cos_half：\n",cos_half)
```

```python
sin_half = sin_table[..., 0::2].unsqueeze(0)
print("sin_half形状：",sin_half.shape)
print("sin_half：\n",sin_half)
```

```python
out_even = x_even * cos_half - x_odd * sin_half
print('out_even形状：',out_even.shape)
print('out_even：\n',out_even)
```

```python
out_odd  = x_even * sin_half + x_odd * cos_half
print('out_odd形状：',out_odd.shape)
print('out_odd：\n',out_odd)
```

```python
stacked = torch.stack([out_even, out_odd], dim=-1)
print('stacked形状：',stacked.shape)
print('stacked:\n',stacked)
```

```python
out = stacked.flatten(-2)
print('out形状：',out.shape)
print('out:\n',out)
```

```python
import torch
import torch.nn as nn

class RoPE(nn.Module):
    def __init__(self, dim: int, max_seq_len: int = 2048):
        """
        初始化函数（厂长建模具）
        dim: 词向量的维度（必须是偶数，比如 64, 128, 4096）
        max_seq_len: 模型支持的最大句子长度
        """
        super().__init__() # 继承 nn.Module 必须写的固定格式
        self.dim = dim
        
        # 1. 计算频率：10000 ^ (-2i / dim)
        # torch.arange(0, dim, 2) 生成偶数索引 [0, 2, 4, ..., dim-2]
        idx = torch.arange(0, dim, 2).float()
        inv_freq = 1.0 / (10000 ** (idx / dim))
        
        # 2. 预计算所有位置的旋转角度表
        # t 是位置 [0, 1, 2, ..., max_seq_len-1]
        t = torch.arange(max_seq_len).float()
        # freqs 形状：[max_seq_len, dim // 2]
        freqs = torch.outer(t, inv_freq)
        
        # 3. 算 cos 和 sin，并两两复制对齐
        # 为什么要 repeat_interleave(2)？因为每两个维度共享一个角度
        # 形状从 [max_seq_len, dim // 2] 变成 [max_seq_len, dim]
        cos_cached = freqs.cos().repeat_interleave(2, dim=-1)
        sin_cached = freqs.sin().repeat_interleave(2, dim=-1)
        
        # 4. 注册为 buffer（极其重要！）
        # 使用 register_buffer 存起来的变量：
        # 1) 不需要计算梯度（不是可训练参数）
        # 2) 会自动跟模型一起搬到 GPU 上（.cuda()）
        # 3) 会保存到模型权重文件（.pth）中
        self.register_buffer("cos_cached", cos_cached)
        self.register_buffer("sin_cached", sin_cached)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        """
        前向传播（工人加工数据）
        x 形状: (batch_size, seq_len, dim)
        """
        # 自动获取当前输入句子的真实长度
        seq_len = x.shape[1]
        
        # 1. 从大表里截取出当前句子长度所需的部分
        # 形状变成: [1, seq_len, dim] (加 unsqueeze(0) 是为了匹配 batch 维度)
        cos = self.cos_cached[:seq_len, :].unsqueeze(0)
        sin = self.sin_cached[:seq_len, :].unsqueeze(0)
        
        # 2. 奇偶拆分
        # x_even 形状: (batch_size, seq_len, dim // 2)
        x_even = x[..., 0::2] 
        x_odd  = x[..., 1::2] 
        
        # 3. cos 和 sin 也截取偶数位，保持形状一致
        # cos_half 形状: [1, seq_len, dim // 2]
        cos_half = cos[..., 0::2]
        sin_half = sin[..., 0::2]
        
        # 4. 核心二维旋转公式
        out_even = x_even * cos_half - x_odd * sin_half
        out_odd  = x_even * sin_half + x_odd * cos_half
        
        # 5. 交错拼装复原
        # out_even 和 out_odd 形状是 (batch_size, seq_len, dim // 2)
        # stack 后变成 (batch_size, seq_len, dim // 2, 2)
        # flatten 后变回 (batch_size, seq_len, dim)
        out = torch.stack([out_even, out_odd], dim=-1).flatten(-2)
        
        return out
    
    # 1. 实例化（创建 RoPE 层）
# 假设我们词向量维度是 6，最大长度支持 10
rope = RoPE(dim=6, max_seq_len=10)

# 2. 造一点假数据（batch_size=1, seq_len=4, dim=6）
x = torch.randn(1, 4, 6)
print("输入形状:", x.shape)

# 3. 使用 RoPE
out = rope(x)
print("输出形状:", out.shape)

# 4. 验证模长守恒（旋转不改变向量长度）
x_norm = torch.norm(x, dim=-1)
out_norm = torch.norm(out, dim=-1)
print("输入向量的长度:\n", x_norm)
print("旋转后向量的长度:\n", out_norm)
```
